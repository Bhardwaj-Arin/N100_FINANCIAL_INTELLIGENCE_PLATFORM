### Module 1 - Scoring Framework

```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:.2f}".format)
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_PATH = PROJECT_ROOT / "data" / "processed" / "master_features.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset Shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
print(df.columns.tolist())
```

```python
score_df = df.copy()

print(score_df.shape)
```

```python
score_columns = [
    "ROE_Score",
    "ROCE_Score",
    "NPM_Score",
    "OPM_Score",
    "ProfitabilityScore",
    "StabilityScore",
    "ValuationScore",
    "MarketScore",
    "FinancialHealthScore",
    "Rating"
]

for col in score_columns:
    score_df[col] = np.nan

score_df.head()
```

### Module 2 - Profitability Score

## Ensure Numeric Columns

```python
profitability_cols = [
    "roe_percentage",
    "roce_percentage",
    "net_profit_margin_pct",
    "operating_profit_margin_pct"
]

for col in profitability_cols:
    score_df[col] = pd.to_numeric(score_df[col], errors="coerce")

score_df[profitability_cols].dtypes
```

## ROE Score

```python
score_df["ROE_Score"] = pd.cut(
    score_df["roe_percentage"],
    bins=[-np.inf, 0, 10, 15, 20, np.inf],
    labels=[0, 25, 50, 75, 100]
).astype(float)
```

## ROCE Score

```python
score_df["ROCE_Score"] = pd.cut(
    score_df["roce_percentage"],
    bins=[-np.inf, 0, 10, 15, 20, np.inf],
    labels=[0, 25, 50, 75, 100]
).astype(float)
```

## Net Profit Margin Score

```python
score_df["NPM_Score"] = pd.cut(
    score_df["net_profit_margin_pct"],
    bins=[-np.inf, 0, 5, 10, 20, np.inf],
    labels=[0, 25, 50, 75, 100]
).astype(float)
```

## Operating Profit Margin Score

```python
score_df["OPM_Score"] = pd.cut(
    score_df["operating_profit_margin_pct"],
    bins=[-np.inf, 0, 5, 10, 20, np.inf],
    labels=[0, 25, 50, 75, 100]
).astype(float)
```

## Calculate Profitability Score

```python
score_df["ProfitabilityScore"] = (
    score_df[
        [
            "ROE_Score",
            "ROCE_Score",
            "NPM_Score",
            "OPM_Score"
        ]
    ]
    .mean(axis=1)
    .round(2)
)
```

## Verifying Results

```python
score_df[
    [
        "company_name",
        "roe_percentage",
        "roce_percentage",
        "net_profit_margin_pct",
        "operating_profit_margin_pct",
        "ROE_Score",
        "ROCE_Score",
        "NPM_Score",
        "OPM_Score",
        "ProfitabilityScore"
    ]
].head(10)
```

### Module 3 - Financial Stability Score

## Convert Required Columns to Numeric

```python
stability_cols = [
    "debt_to_equity",
    "asset_turnover",
    "free_cash_flow_cr",
    "cash_from_operations_cr"
]

for col in stability_cols:
    score_df[col] = pd.to_numeric(score_df[col], errors="coerce")

score_df[stability_cols].dtypes
```

## Debt-to-Equity Score

```python
score_df["DebtEquity_Score"] = pd.cut(
    score_df["debt_to_equity"],
    bins=[-np.inf, 0.5, 1, 2, 3, np.inf],
    labels=[100, 75, 50, 25, 0]
).astype(float)
```

## Asset Turnover Score

```python
score_df["AssetTurnover_Score"] = pd.cut(
    score_df["asset_turnover"],
    bins=[-np.inf, 0.5, 1, 1.5, 2, np.inf],
    labels=[0, 25, 50, 75, 100]
).astype(float)
```

## Free Cash Flow Score

```python
score_df["FCF_Score"] = np.where(
    score_df["free_cash_flow_cr"] > 0,
    100,
    0
)
```

## Operating Cash Flow Score

```python
score_df["OperatingCashFlow_Score"] = np.where(
    score_df["cash_from_operations_cr"] > 0,
    100,
    0
)
```

## Final Stability Score

```python
score_df["StabilityScore"] = (
    score_df[
        [
            "DebtEquity_Score",
            "AssetTurnover_Score",
            "FCF_Score",
            "OperatingCashFlow_Score"
        ]
    ]
    .mean(axis=1)
    .round(2)
)
```

## Outputs

```python
score_df[
    [
        "company_name",
        "debt_to_equity",
        "asset_turnover",
        "free_cash_flow_cr",
        "cash_from_operations_cr",
        "DebtEquity_Score",
        "AssetTurnover_Score",
        "FCF_Score",
        "OperatingCashFlow_Score",
        "StabilityScore"
    ]
].head(10)
```

### Module 4 - Valuation Score

## Convert Valuation Columns to Numeric

```python
valuation_cols = [
    "pe_ratio",
    "pb_ratio",
    "ev_ebitda",
    "dividend_yield_pct",
    "market_cap_crore"
]

for col in valuation_cols:
    score_df[col] = pd.to_numeric(score_df[col], errors="coerce")

score_df[valuation_cols].dtypes
```

## PE Ratio Score

```python
score_df["PE_Score"] = pd.cut(
    score_df["pe_ratio"],
    bins=[-np.inf, 15, 25, 40, 60, np.inf],
    labels=[100, 75, 50, 25, 0]
).astype(float)
```

## PB Ratio Score

```python
score_df["PB_Score"] = pd.cut(
    score_df["pb_ratio"],
    bins=[-np.inf, 1, 3, 5, 10, np.inf],
    labels=[100, 75, 50, 25, 0]
).astype(float)
```

## EV/EBITDA Score

```python
score_df["EVEBITDA_Score"] = pd.cut(
    score_df["ev_ebitda"],
    bins=[-np.inf, 8, 12, 18, 25, np.inf],
    labels=[100, 75, 50, 25, 0]
).astype(float)
```

## Dividend Yield Score

```python
score_df["DividendYield_Score"] = pd.cut(
    score_df["dividend_yield_pct"],
    bins=[-np.inf, 0, 1, 2, 4, np.inf],
    labels=[0, 25, 50, 75, 100]
).astype(float)
```

## Final Valuation Score

```python
score_df["ValuationScore"] = (
    score_df[
        [
            "PE_Score",
            "PB_Score",
            "EVEBITDA_Score",
            "DividendYield_Score"
        ]
    ]
    .mean(axis=1)
    .round(2)
)
```

## Verifying Results

```python
score_df[
    [
        "company_name",
        "pe_ratio",
        "pb_ratio",
        "ev_ebitda",
        "dividend_yield_pct",
        "PE_Score",
        "PB_Score",
        "EVEBITDA_Score",
        "DividendYield_Score",
        "ValuationScore"
    ]
].head(10)
```

### Module 5 - Market Score

## Convert Market Columns

```python
market_cols = [
    "market_cap_crore",
    "sector_rank",
    "sector_company_count"
]

for col in market_cols:
    score_df[col] = pd.to_numeric(score_df[col], errors="coerce")

score_df[market_cols].dtypes
```

## Market Cap Score

```python
score_df["MarketCap_Score"] = pd.cut(
    score_df["market_cap_crore"],
    bins=[
        -np.inf,
        5000,
        20000,
        100000,
        500000,
        np.inf
    ],
    labels=[
        20,
        40,
        60,
        80,
        100
    ]
).astype(float)
```

## Sector Rank Score

```python
score_df["SectorRank_Score"] = (
    (
        score_df["sector_company_count"]
        - score_df["sector_rank"]
    )
    /
    (
        score_df["sector_company_count"] - 1
    )
) * 100

score_df["SectorRank_Score"] = (
    score_df["SectorRank_Score"]
    .clip(0, 100)
    .fillna(50)
    .round(2)
)
```

## Final Market Score

```python
score_df["MarketScore"] = (
    score_df[
        [
            "MarketCap_Score",
            "SectorRank_Score"
        ]
    ]
    .mean(axis=1)
    .round(2)
)
```

## Verifying Results

```python
score_df[
    [
        "company_name",
        "market_cap_crore",
        "sector_rank",
        "sector_company_count",
        "MarketCap_Score",
        "SectorRank_Score",
        "MarketScore"
    ]
].head(10)
```

### Module 6 - Composite Financial Health Score

## Calculate Financial Health Score

```python
score_df["FinancialHealthScore"] = (
    score_df["ProfitabilityScore"] * 0.35
    + score_df["StabilityScore"] * 0.30
    + score_df["ValuationScore"] * 0.20
    + score_df["MarketScore"] * 0.15
).round(2)
```

## Company Rating

```python
score_df["Rating"] = pd.cut(
    score_df["FinancialHealthScore"],
    bins=[0, 40, 55, 70, 80, 90, 100],
    labels=["D", "C", "B", "B+", "A", "A+"],
    include_lowest=True
)
```

## Company Ranking

```python
score_df["OverallRank"] = (
    score_df["FinancialHealthScore"]
    .rank(method="dense", ascending=False)
    .astype(int)
)
```

## Sector-wise Ranking

```python
score_df["SectorRankFinal"] = (
    score_df
    .groupby("broad_sector")["FinancialHealthScore"]
    .rank(method="dense", ascending=False)
    .astype(int)
)
```

## Verifying Results

```python
score_df[
    [
        "company_name",
        "broad_sector",
        "ProfitabilityScore",
        "StabilityScore",
        "ValuationScore",
        "MarketScore",
        "FinancialHealthScore",
        "Rating",
        "OverallRank",
        "SectorRankFinal"
    ]
].sort_values(
    by="FinancialHealthScore",
    ascending=False
).head(20)
```

### Module 7 - Company & Sector Rankings

## Top 20 Companies Overall

```python
top_companies = (
    score_df[
        [
            "OverallRank",
            "company_name",
            "broad_sector",
            "FinancialHealthScore",
            "Rating"
        ]
    ]
    .sort_values("OverallRank")
    .drop_duplicates(subset="company_name")
    .head(20)
)

top_companies
```

## Top 5 Companies in Each Sector

```python
top_sector_companies = (
    score_df[
        [
            "company_name",
            "broad_sector",
            "FinancialHealthScore",
            "SectorRankFinal",
            "Rating"
        ]
    ]
    .sort_values(
        ["broad_sector", "SectorRankFinal"]
    )
    .drop_duplicates(subset=["broad_sector", "company_name"])
    .groupby("broad_sector")
    .head(5)
)

top_sector_companies
```
